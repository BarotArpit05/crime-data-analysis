# Crime Data Analysis in India using Python

This project analyzes crime data across different states of India for the years 2020, 2021, and 2022. The purpose of this analysis is to understand state-wise crime patterns, compare trends over the years, and identify useful insights from the available data.

The analysis is carried out using Python libraries such as NumPy, Pandas, and Matplotlib in Jupyter Notebook. The project includes data cleaning, transformation, analysis, and visualization.

## Dataset Selection

The dataset selected for this project contains state-wise crime statistics of India. It includes information such as total IPC crimes for the years 2020, 2021, and 2022, population estimates for 2022, crime rate, and chargesheeting rate.

This dataset was selected because it contains multiple variables that allow comparison, trend analysis, and relationship study. It is suitable for exploratory data analysis because it combines numerical values across different years and regions.

## Data Cleaning

Data cleaning is performed to ensure accuracy of analysis.

The following steps are applied:

### 1. Missing Value Check

The isnull().sum() method is used to identify null values in each column. Missing values can affect calculations and visual outputs.

### 2. Duplicate Check

The duplicated().sum() method is used to check repeated rows. Duplicate records may create incorrect analysis.

### 3. Duplicate Removal

The drop_duplicates() method removes repeated entries.

### 4. Removing Unnecessary Column

The serial number column is removed because it does not contribute to analysis. It is only an index value.

### 5. Renaming Columns

Long column names are renamed to shorter names to improve readability and simplify coding.

### 6. Data Type Verification

The dtypes method is used to verify whether numerical columns are stored correctly. This ensures calculations can be performed without errors.

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.drop("Sl. No.",axis=1,inplace=True)
```

```python
df.columns = ['State','Crime_2020','Crime_2021','Crime_2022',
              'Population_2022','Crime_Rate_2022','Chargesheet_Rate']
```

```python
df.dtypes
```

## Feature Engineering

Additional columns are created to improve analysis.

### Total Crime

A new column is created by adding crime values of 2020, 2021, and 2022. This gives the combined crime count for each state.

### Average Crime

A second column is created to calculate average crime across the three years.

These features make comparison easier and provide a broader understanding of long-term crime patterns.

```python
df['Total_Crime']= df['Crime_2020']+df['Crime_2021']+df['Crime_2022']
```

```python
df['Avg_Crime'] = df[['Crime_2020','Crime_2021','Crime_2022']].mean(axis=1)
```

## Exploratory Data Analysis

The cleaned dataset is analyzed to identify important patterns.

### Top States

States are sorted based on total crime in descending order. This helps identify the highest crime regions.

### Low Crime States

States are sorted in ascending order to identify regions with lower crime counts.

Sorting helps in ranking states and preparing the dataset for visualizations.

```python
top_states = df.sort_values(by='Total_Crime', ascending=False)
top_states.head(10)
```

```python
low_states = df.sort_values(by='Total_Crime')
low_states.head(10)
```

## Visualization 1: Top 10 States by Total Crime

A bar chart is used to compare total crime among the top 10 states.

### Method Used

bar()

### Reason

Bar charts are suitable when comparing values across categories.

### Output

The chart shows which states have the highest combined crime counts over three years.

### Important Formatting

- figure(figsize) increases graph size for readability.
- xticks(rotation=60) rotates state names to avoid overlap.
- xlabel and ylabel add axis labels.
- title provides chart description.

### Observation

This visualization makes it easy to identify the states contributing the highest crime totals.

```python
top10=df.sort_values(by='Total_Crime',ascending=False).head(10)
plt.figure(figsize=(12,6))
plt.grid(True)
plt.bar(top10['State'],top10['Total_Crime'])
plt.xticks(rotation=60)
plt.title("Top 10 States By Total Crime")
plt.xlabel("State")
plt.ylabel("Total Crime")
plt.tight_layout()
plt.show()
```

## Visualization 2: Year-wise Crime Trend

A line chart is used to study changes in total crime across years.

### Method Used

plot()

### Reason

Line charts are useful for showing trends over time.

### Output

The chart displays whether total crime increased or decreased between 2020 and 2022.

### Important Formatting

- marker='o' highlights each year’s data point.
- index is used for x-axis labels.
- values are used for total crime counts.

### Observation

This visualization helps in understanding annual changes in crime.

```python
year_total=df[['Crime_2020','Crime_2021','Crime_2022']].sum()
plt.figure(figsize=(10,5))
plt.plot(year_total.index,year_total.values,marker='o',color='red')
plt.grid(True)
plt.title("Crime Trend in India (2020-2022)")
plt.xlabel("Year")
plt.ylabel("Total Crimes")
plt.show()
```

## Visualization 3: Population vs Crime

A scatter plot is created to study the relationship between population and crime.

### Method Used

scatter()

### Reason

Scatter plots are useful to compare two numerical variables.

### Output

Each point represents one state.

X-axis shows population.

Y-axis shows crime.

### Observation

This chart helps determine whether states with larger populations have higher crime levels.

```python
plt.figure(figsize=(10,6))
plt.scatter(df['Population_2022'],df['Crime_2022'])
plt.grid(True)
plt.xlabel("Population")
plt.ylabel("Crime")
plt.title("Population VS Crime in Year 2022")
plt.show()
```

## Visualization 4: Crime Share of Top 5 States

A pie chart is used to show the percentage contribution of the top 5 states.

### Method Used

pie()

### Reason

Pie charts are useful for showing proportions.

### Output

The chart shows the percentage share of total crime among selected states.

### Important Formatting

- labels shows state names
- autopct displays percentage values
- figure size ensures proper display

### Observation

This chart helps compare contribution share among the top states.

```python
top5 = df.sort_values(by='Total_Crime', ascending=False).head(5)

plt.figure(figsize=(8,8))
plt.pie(top5['Total_Crime'], labels=top5['State'], autopct='%1.1f%%')
plt.title("Crime Share of Top 5 States")
plt.show()
```

## Visualization 5: Box Plot

A box plot is used to identify spread and outliers in crime values.

### Method Used

boxplot()

### Reason

It provides information about median, range, and extreme values.

### Output

The chart highlights unusual states with significantly higher crime.

### Observation

This helps detect outliers in the dataset.

```python
plt.figure(figsize=(8,6))
plt.boxplot(df['Crime_2022'])
plt.title("Box Plot of Crime 2022")
plt.ylabel("Crime Cases")
plt.show()
```

## Visualization 8: Multi-Line Comparison

A multi-line chart compares crime for 2020, 2021, and 2022 for all states.

### Method Used

plot() with multiple lines

### Reason

This allows year-by-year comparison in one graph.

### Output

Three lines represent three years.

### Observation

This helps compare yearly changes at state level.

```python
plt.figure(figsize=(12,6))

plt.plot(df['State'], df['Crime_2020'], label='2020')
plt.plot(df['State'], df['Crime_2021'], label='2021')
plt.plot(df['State'], df['Crime_2022'], label='2022')

plt.xticks(rotation=90)
plt.grid(True)
plt.legend()
plt.title("Crime Comparison by Year")
plt.xlabel("State")
plt.ylabel("Crime")
plt.show()
```
